```python
def cos_sim(x,y):
    return np.dot(x,y)/(np.linalg.norm(x)*np.linalg.norm(y))
def l2(x,y):
    return np.linalg.norm(x-y)
def compare2CosSim(h1,h2):
    return np.array([
        l2(h1[j],h2[j]) for j in range(h2.shape[0])
    ])


compare2CosSim(
    all_shap_values_dict_loaded["016_iid"]["Server"],
    all_shap_values_dict_loaded["016_imb_0.99"]["Server"]
    ).mean()
```
